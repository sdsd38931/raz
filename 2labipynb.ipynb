```python

!pip install gdown

import gdown
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import cross_val_score
import numpy as np


gdown.download('https://drive.google.com/uc?id=1k3SXSzlgwdlpEfI6qbqGZgR-uycfpKl8', 'SMSSpamCollection.txt', quiet=False)


data = pd.read_csv('SMSSpamCollection.txt', sep='\t', header=None, names=['label', 'message'])
texts = data['message'].tolist()
labels = [1 if label == 'spam' else 0 for label in data['label']]


vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)


log_reg = LogisticRegression(max_iter=1000, random_state=2)
f1_scores = cross_val_score(log_reg, X, labels, cv=10, scoring='f1')
mean_f1_score = np.mean(f1_scores)

print(f'Mean F1 Score (Logistic Regression): {mean_f1_score:.1f}')


log_reg.fit(X, labels)

test_messages = [
    "FreeMsg: Txt: CALL to No: 86888 & claim your reward of 3 hours talk time to use from your phone now! Subscribe6GB",
    "FreeMsg: Txt: claim your reward of 3 hours talk time",
    "Have you visited the last lecture on physics?",
    "Have you visited the last lecture on physics? Just buy this book and you will have all materials! Only 99$",
    "Only 99$"
]

test_X = vectorizer.transform(test_messages)
predictions = log_reg.predict(test_X)

print("Predictions (Logistic Regression):", ' '.join(map(str, predictions)))


for ngram in [(2, 2), (3, 3), (1, 3)]:
    vectorizer_ngram = CountVectorizer(ngram_range=ngram)
    X_ngram = vectorizer_ngram.fit_transform(texts)
    f1_scores_ngram = cross_val_score(log_reg, X_ngram, labels, cv=10, scoring='f1')
    mean_f1_ngram = np.mean(f1_scores_ngram)
    print(f'Mean F1 Score (ngram_range={ngram}): {mean_f1_ngram:.2f}')


for ngram in [(2, 2), (3, 3), (1, 3)]:
    vectorizer_ngram = CountVectorizer(ngram_range=ngram)
    X_ngram = vectorizer_ngram.fit_transform(texts)
    nb_classifier = MultinomialNB()
    f1_scores_nb = cross_val_score(nb_classifier, X_ngram, labels, cv=10, scoring='f1')
    mean_f1_nb = np.mean(f1_scores_nb)
    print(f'Mean F1 Score (MultinomialNB with ngram_range={ngram}): {mean_f1_nb:.2f}')


tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(texts)
f1_scores_tfidf = cross_val_score(log_reg, X_tfidf, labels, cv=10, scoring='f1')
mean_f1_tfidf = np.mean(f1_scores_tfidf)

print(f'Mean F1 Score (Tfidf + Logistic Regression): {mean_f1_tfidf:.2f}')


if mean_f1_tfidf > mean_f1_score + 0.01:
    print("Result: 1")
elif mean_f1_tfidf < mean_f1_score - 0.01:
    print("Result: -1")
else:
    print("Result: 0")
```
